```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

driver = webdriver.Chrome()
driver.maximize_window()

try:
    driver.get("https://www.google.com")
    search_box = driver.find_element(By.NAME, "q")
    search_box.send_keys("Amazon" + Keys.RETURN)
    time.sleep(10)

    amazon_link = driver.find_element(By.XPATH, "//h3[contains(text(),'Amazon')]")
    amazon_link.click()
    time.sleep(3)

    amazon_search = driver.find_element(By.ID, "twotabsearchtextbox")
    amazon_search.send_keys("laptops")
    search_button = driver.find_element(By.ID, "nav-search-submit-button")
    search_button.click()
    time.sleep(3)

    driver.refresh()
    time.sleep(2)

    product_titles = driver.find_elements(
        By.CLASS_NAME, "a-size-medium.a-color-base.a-text-normal"
    )

    print("\n--- Extracted Product Titles ---")
    for index, title in enumerate(product_titles[:5], start=1):
        print(f"{index}. {title.text}")

finally:
    driver.quit()
```
